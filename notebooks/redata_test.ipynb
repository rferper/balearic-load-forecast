```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/balearic_demand_raw.csv")
df["datetime"] = pd.to_datetime(df["datetime"], utc=True)
df.info()
```

```python
start = pd.Timestamp("2026-08-01", tz="UTC")   # tz="UTC" matters here - see note below
end = pd.Timestamp("2026-08-27", tz="UTC")

subset = df[(df["datetime"] >= start) & (df["datetime"] < end)]

subset.plot(x="datetime", y="value", figsize=(10, 4))
plt.ylabel("Demand (MW)")
plt.title(f"Balearic hourly demand — {start.date()} to {end.date()}")
plt.show()
```
